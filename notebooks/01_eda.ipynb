# 1. Exploratory Data Analysis (EDA)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## 1.1 Load Dataset & Kiểm tra cơ bản

```python
df = pd.read_csv("../data/heart_cleveland_upload.csv")
print("Shape:", df.shape)
df.head()
```

```python
df.info()
```

```python
df.nunique()
```

## 1.2 Descriptive Statistics & Missing values

```python
print("Missing values:\n", df.isnull().sum())
print("Duplicates:", df.duplicated().sum())
```

```python
df.describe()
```

## 1.3 Target Distribution (Phân phối nhãn mục tiêu)

```python
sns.countplot(data=df, x="condition")
plt.title("Target Distribution (Condition)")
plt.show()
```

## 1.4 Correlation Analysis (Phân tích tương quan)

```python
plt.figure(figsize=(12, 10))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Feature Correlation")
plt.show()
```

## 1.5 Nhận xét
- Dữ liệu không chứa missing values.
- Biến mục tiêu khá cân bằng (class balance).
- Có sự tương quan nhẹ giữa một số biến như `thalach` và `age`.
